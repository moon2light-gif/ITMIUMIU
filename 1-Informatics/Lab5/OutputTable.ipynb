```python
import pandas as pd
from openpyxl.reader.excel import load_workbook
from openpyxl.styles import *

file_in = "Крамарь ИнфоЛаб5_right.ods"
# прочитаем мой файл.ods

pd.set_option('display.max_rows', 100)
pd.set_option('display.max_columns', 25) # настраиваем максимальное количество столбцов
pd.set_option('display.width', 1000)
pd.set_option('display.max_colwidth', 100)

df = pd.read_excel(file_in, engine="odf", header=None)
# в моём сформированном файле есть заголовки, поэтому убираем их

# первое - срез строк, второе - столбцов
df_cut = df.iloc[3:15, 0:25]

# вырезаем столбец F
df_cut = df_cut.drop(df_cut.columns[5], axis=1)

# заменяем NaN на пустые строки
df_cut = df_cut.fillna("")

df_processed = df_cut.astype(object)

# идем построчно
for row in df_processed.index:
    # идём по каждой колонке
    for col in df_processed.columns:
        value = df_processed.loc[row, col]
        if isinstance(value, float):
            df_processed.loc[row, col] = str(value)[:-2]

def style_column(s):
    styles = []
    column_name = df_processed.columns.get_loc(s.name) # получаем номер столбца
    for i in range(len(s)):
        left = "2px solid blue"
        right = "2px solid blue"
        top = "none"
        bottom = "none"

        if column_name == 4:
            left = "none"

        elif column_name == 3:
            right = "none"

        elif 5 <= column_name <= 22:
            left = "none"
            right = "none"

        elif column_name == 23:
            left = "none"
            right = "2px solid blue"

        if i == 0:
            top = "2px solid blue"

        elif i == len(s) - 1:
            bottom = "2px solid blue"

        style = (
           f"border-left: {left};"
           f"border-right: {right};"
           f"border-top: {top};"
           f"border-bottom: {bottom};"
        )

        styles.append(style)

    return styles

styled_df = df_processed.style.apply(style_column, subset=df_processed.columns[:24])

# вывод DataFrame в виде таблицы в Jupyter Notebook
styled_df


```
